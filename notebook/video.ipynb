```python
import boto3
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
# Let's use Amazon S3
s3 = boto3.resource(service_name = 's3' )
s3_client = boto3.client("s3")
```

```python
# Print out bucket names
for bucket in s3.buckets.all():
    print(bucket.name)
```

```python
client_name = "test_client_1"
cam_name = "test_cam_1"
object_key  = f"abc/{client_name}/{cam_name}/"
video_date = "20250607"
```

```python
for obj in s3.Bucket("testing-cvat-load").objects.filter(Prefix = object_key):
    path, filename = os.path.split(obj.key)
    print(path , " ",filename)
    if video_date in filename:
        print("We found video" , filename)
        os.makedirs("temm_video")
        s3_client.download_file(
                Filename=f"{"temm_video"}/{filename}",
                Bucket="testing-cvat-load",
                Key=obj.key
        )
        
```
